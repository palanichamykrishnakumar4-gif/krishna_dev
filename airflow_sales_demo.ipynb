```python
from datetime import datetime

data = [
    (101, "Laptop", 55000),
    (102, "Mobile", 25000),
    (103, "Tablet", 18000),
    (104, "Monitor", 12000),
    (105, "Keyboard", 2000)
]

columns = ["sale_id", "product", "amount"]

df = spark.createDataFrame(data, columns)

# Calculate total sales
df.createOrReplaceTempView("sales_temp")

total_sales = spark.sql("""
    SELECT SUM(amount) AS total_sales
    FROM sales_temp
""")

display(df)

print("Total Sales:")
display(total_sales)

# Save output
df.write.mode("overwrite").saveAsTable(
    "test_cate.test_schema.mytable1"
)

print("Airflow Databricks Demo Completed")
```
